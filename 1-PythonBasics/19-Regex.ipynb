```python
# Regular expressions allows us to find for certain patterns of text or numbers

# to work with regular expressions in python we have to import 're' package

import re

# raw string in python is a string prefixed with r which tell python not 
# to handle \ in a special way

print("\tTab")
print(r"\tTab")
```

```python
# finditer() gathers all the matches in the text

text_to_serach = """ABCDEFGHIJKLMNOPQRSTUVWXYZ
abcdefghijklmnopqrstuvwxyz
12345678910
Later that day, rohan emailed a reminder:
“PLEASE VERIFY THE ACCESS CODES: abC123xyZ, qRt890LMn.”

Ha HaHa
abc egf abc
Meanwhile, EMILY updated the contact sheet:

David Carter — 214-789-9080

linda rogers — 469-555-9422

PETER HAMILTON — 830-555-7744



"""
# finging specific lietrals in text

pattern = re.compile(r'abc') 

matches = pattern.finditer(text_to_serach) # finding the pattern in string

for match in matches:
    print(match)

print(text_to_serach[27:30])
```

```python
pattern = re.compile(r'-')

matches = pattern.finditer(text_to_serach)

for match in matches:
    print(match)
```

```python
# meta characters that need to be escaped while finding pattern  are 
# . ^ $ * + ? { } [ ] ( )  \  |

pattern = re.compile(r'\.')
matches = pattern.finditer(text_to_serach)

for match in matches:
    print(match)
```

```python
'''
finding the patterns in text

 .  -> any character except new line
 \d -> Digit (0-9)
 \D -> Not a Digit (0-9)
 \w -> word character (a-z A-Z 0-9 _)
 \W -> Not a word character 
 \s -> white space (space tab newline)
 \S -> Not white space (space tab newline)

 \b -> word boundary
 \B -> Not a word boundary
 ^ -> Beginning of String
 $ -> End of String
 
 
'''
```

```python
pattern = re.compile(r'\W')
matches = pattern.finditer(text_to_serach)
for match in matches:
    print(match)
```

```python
pattern = re.compile(r'\bHa') # finds the pattern where the word starts with Ha
matches = pattern.finditer(text_to_serach)
for match in matches:
    print(match)
```

```python
pattern = re.compile(r"word$")
matches = pattern.finditer(sentence)
for match in matches:
    print(match)
```

```python
# matches all the phone numbers seperated with any symbol
# pattern = re.compile(r'\d\d\d.\d\d\d.\d\d\d\d')
# suppose we want phone numbers seperated with only . or -
# the characters in [] looks for only 1 character match in the square brackets
# pattern = re.compile(r'\d\d\d[.-]\d\d\d[.-]\d\d\d\d')
# to find phone numbers that start with 800 and 900
pattern = re.compile(r'[89]00[.-]\d\d\d[.-]\d\d\d\d')
with open("contacts.txt",'r',encoding='utf-8') as f:
    contents = f.read()
    matches = pattern.finditer(contents)
for match in matches:
    print(match)

```

```python
#find the phone numbers that start with 1-5
pattern = re.compile(r'[1-5]\d\d[.-]\d\d\d[.-]\d\d\d\d')
with open("contacts.txt",'r',encoding='utf-8') as f:
    contents = f.read()
    matches = pattern.finditer(contents)

for match in matches:
    print(match)    
```

```python
sentence = "Starts with 0 and ends with the string"

# when ^ is placed in character set , then it means it negates the set

pattern = re.compile(r'[^A-Za-z]') # matches all characters that are not alphabets
matches = pattern.finditer(sentence)
for match in matches:
    print(match)   


```

```python
# match all the words like pat cat mat except bat

list_of_items = " cat mat pot bat sat satman"


pattern = re.compile(r'[^b]at ')

matches = pattern.finditer(list_of_items)

for match in matches:
    print(match)   

```

```python
# using quantifiers to match multiple characters at a time
'''
* -> 0 or more
+ -> 1 or more
? -> 0 or 1
{3} -> exact number
{3,4} -> range of numbers (min,max)

'''

# pattern to find the valid phone numbers using quantifiers

pattern=re.compile(r'\d{3}[.-]\d{3}[.-]\d{4}')
with open("contacts.txt",'r') as f:
    contents  = f.read()
    matches = pattern.finditer(contents)

for match in matches:
    print(match)   
```

```python
name_search = '''
Mr. David
Mrs. David
Mr. T
Mr. Chaitanya
Mr Corey
Ms Alekhya
'''
```

```python
# find the names that start with Mr

pattern = re.compile(r'Mr\.?\s[A-Z]\w*')
matches = pattern.finditer(name_search)

for match in matches:
    print(match)  
```

```python
# using groups u can match several pattern
# now match the names that all start with Mr, Mrs,Ms

pattern = re.compile(r'M(r|rs|s)\.?\s[A-Z]\w*')
matches = pattern.finditer(name_search)

for match in matches:
    print(match)  
```

```python
emails = '''

alekhya@gmail.com
sri@gmail.co
lakshmi.k@university.edu
chay-345@my_work.net
'''
```

```python
urls = '''
    https://www.google.com
    http://coreyms.com
    https://youtube.com
    https://www.nasa.gov
'''
```

```python
# here we try to access the information from urls using groups ()

pattern = re.compile(r'https?://(www\.)?(\w+)(\.\w+)')

matches = pattern.finditer(urls)

for match in matches:
    print(match.group(0)) # group 0 is original url
    print(match.group(1)) # group 1 is first group www.
    print(match.group(2))
    print(match.group(3))
    
    

```

```python
# formating the urls using groups

pattern = re.compile(r'https?://(www\.)?(\w+)(\.\w+)')
sub_urls = pattern.sub(r'\2\3',urls) # includes groups 2 and 3

print(sub_urls)
```

```python
# we can use findall instead of finditer
'''
difference between findall and finditer

using finditer 
You can get:
.group() → matched text
.start() → start index
.end() → end index
.span() → (start, end)

using findall you get: 
Output is only the matched text
Easy when you just need the matches
No info about where the match occurred

'''

# pattern = re.compile(r'https?://(www\.)?(\w+)(\.\w+)') # returns group wise
pattern = re.compile(r'https?://www\.?\w+\.\w+')

matches = pattern.findall(urls)

for match in matches:
    print(match)  
```

```python
# match -> returns if the match is present at start of String, if the match 
# is in middle it returns None

sentence = "Starts with 0 and ends with the string"

# pattern = re.compile(r'Start')
pattern = re.compile(r'with')
matches  = pattern.match(sentence)
print(matches)
```

```python
# search will return first match if the match is present anywhere in string

pattern = re.compile(r'with')
matches = pattern.search(sentence)
print(matches)
```

```python
# using flags
# suppose we want to find the match of specific word irrespective of case
# we use IGNORECASE flag or I as short form

pattern = re.compile(r'start',re.IGNORECASE)
match = pattern.search(sentence)
print(match)
```
