```python
# closures

'''
A closure is created when:
You have a nested function (a function inside another function),
The inner function uses variables from the outer function,
The outer function returns the inner function.
The returned inner function keeps the outer function’s variables alive.
'''

def main_welcome():
    msg = "welcome"
    def sub_welcome_method():
        print("wlecome to advanced python course")
        print("please learn the concepts properly")
        print(msg)
    return sub_welcome_method()

```

```python
main_welcome()
```

```python
def main_welcome(func):
    def sub_welcome_func():
        print("wlecome to advanced python course")
        print("please learn the concepts properly")
        func("this is from function")
    return sub_welcome_func()
```

```python
main_welcome(print)
```

```python
def main_welcome(func,lst):
    def sub_welcome_func():
            print("wlecome to advanced python course")
            print("please learn the concepts properly")
            print(func(lst))
    return sub_welcome_func()
```

```python
main_welcome(len,[1,2,3,4,5,6])
```

```python
main_welcome(min,[1,2,3,4,5,6])
main_welcome(max,[1,2,3,4,5,6])
```

```python
# Decorator
# without using decorators
def main_welcome(func):
    def sub_welcome_func():
            print("wlecome to advanced python course")
            print("please learn the concepts properly")
            func()
    return sub_welcome_func()

def course_intro():
      print("this is adv python course")
```

```python
main_welcome(course_intro)
```

```python
# using decorators
@main_welcome # when using @main_welcome the course_intro
# is called in main_welcome
def course_intro():
      print("this is adv python course")
```

```python
def my_decorator(func):
    def wrapper():
        print("before function call")
        func()
        print("after function call")
    return wrapper()

@my_decorator
def say_hello():
    print("say hello from my decorator")
```

```python
import time
def timer(func):
    def wrapper(*args,**kwargs):
        start = time.time()
        result = func(*args,**kwargs)
        end = time.time()
        time_taken = (end -start)*1000
        print(f"function {func.__name__} took {time_taken} seconds")
        return result   
    return wrapper
```

```python
@timer
def find_squares(lst):
    result = []
    for i in lst:
        result.append(i**2)
    return result
```

```python
result = find_squares(range(10000))
```
