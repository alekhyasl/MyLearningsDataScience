```python
# To do list

to_do_list = ['Groceries','Clean house','Pay bills']

# Adding task to list
to_do_list.append('Schedule meeting')
to_do_list.append("Go for run")
print(to_do_list)
```

```python
# Remove completed task
to_do_list.remove('Clean house')
print(to_do_list)
```

```python
# checking if task in list
if 'Pay bills' in to_do_list:
    print("Do not forget to pay bills")
```

```python
# pring remainiing to_do_list

for task in to_do_list:
    print(task)
```

```python
# create a list to store and calculate average grade for students

grades = [89,76,98,67,90,69]

# adding new grades

grades.append(65)

# calculate average grade

avg_grade = sum(grades)/len(grades)
print(f"avegrage grades are : {avg_grade}")

# find highest and lowest grade

print(f"lowest grade : {min(grades)}")

print(f"highest grade : {max(grades)}")
```

```python
# managing inventory in grocery store

inventory = ["Apples","mangoes","Kiwi","oranges"]

# adding new item

inventory.append("guvava")
print(inventory)

# remove out of stock item

inventory.remove("mangoes")
print(inventory)
```

```python
# check if item is in stock

item = "oranges"

if item in inventory:
    print("item in stock")
else:
    print("item out of stock")
```

```python
# use list to analyse and collect user feedback from facebook

feedback = ['Great Service',"Very Satisfied","Could Be Better","Excellent Experience"]

# Adding new feedback

feedback.append("Not Happy With The Service")

# counting the positive feedback
count = 0
for comment in feedback:
     if "great" in comment.lower() or "satisfied" in comment.lower() or "excellent" in comment.lower():
          count += 1

print(f" count of positive feedback :{count}")

# print all the feedback

for comment in feedback:
     print(comment)
```
