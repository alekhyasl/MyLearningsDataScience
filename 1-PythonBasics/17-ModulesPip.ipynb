```python
# when you create an env not all the package will be installed 
# you have install them explicitly
import numpy as np


# q1 and q2 are quarterly results of a company wrt to its products in
# different countries
# row represent country
# column represent product
```

```python
q1 = np.array([
    [200,220,250], # product a
    [150,180,210], # product b
    [300,330,360]  # product c
])

q2 = np.array([
    [209,231,259], # product a
    [155,192,222], # product b
    [310,340,375]  # product c
])
```

```python
# total sales in both quarters

qsum = q1 + q2
print(qsum)
```

```python
qinc = q2 - q1
print(qinc)
```
