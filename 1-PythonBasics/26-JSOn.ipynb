```python
# JSON -> Javascript object Notation
# JSON is the data exchange format which is used by most if software system
# in the world
# There is other format called XML
# In XML we see the name format is repeated, in jSON it is not
# so JSON is called light weight format

'''
JSON format

{
    "name" : "tom"
    "address" : "4201 Monterey oaks blvd"
    "phone" : "7876677890"
}

XML Format

<name>tom</name>
<address>4201 Monterey oaks blvd</address>
<phone>7876677890</phone>

'''
```

```python
# JSON string
JSON_string = '''{
  "name": "Alekhya",
  "age": 33,
  "is_employee": true,
  "skills": ["Python", "Data Analysis", "Machine Learning"],
  "address": {
    "street": "Monterey Oaks Blvd",
    "city": "Austin",
    "zipcode": "78749"
  }
}'''
```

```python
print(type(JSON_string))
```

```python
import json
record = json.loads(JSON_string)
print(record)
```

```python
print(type(record))
```

```python
record['skills']
```

```python
json_str = json.dumps(record)
print(json_str)
```

```python
print(type(json_str))
```

```python
'''
json.dumps() → returns a JSON string

json.dump() → writes JSON directly to a file
'''
with open("output.json",'w') as f:
    json.dump(record,f)
```

```python
json_book_str = '''
{
  "books": [
    {
      "title": "The Silent River",
      "author": "Maya Thompson",
      "published_year": 2018
    },
    {
      "title": "Echoes of Tomorrow",
      "author": "Daniel Carter",
      "published_year": 2020
    },
    {
      "title": "Shadows of the Library",
      "author": "Priya Raman",
      "published_year": 2015
    },
    {
      "title": "Journey Beyond the Stars",
      "author": "Liam Anderson",
      "published_year": 2022
    }
  ]
}

'''
```

```python
# get the title and author of the books in above json_str

book_json = json.loads(json_book_str)
print(book_json)
```

```python
book_list = book_json['books']
```

```python
[{'title':book['title'],'author':book['author']} for book in book_list]
```

```python
#Many times we get a json string which is mal formed
json_info = '''
    "name": "Alekhya",
      "age": 33
      "address":"4201 Monterey Oaks blvd"
'''
# when we try to load this mal formed string we get JSONDecodeError 
# which needs to be handled 
# it is good practice to handle error with try except
try:

    info = json.loads(json_info)
except json.JSONDecodeError as ex:
    print("mal formed string: ",ex)
```
