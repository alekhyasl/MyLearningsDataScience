```python
# determine if a year is a leap year or not
year = int(input("enter year"))

if year % 4 ==0:
    if year%100==0:
        if year%400==0 :
            print(f"{year} is a leap year")
        else:
            print(f"{year} is not a leap year")
    else:
        print(f"{year} is a leap year")
    
else:
    print(f"{year} is not a leap year")
```

```python
# write a code for simple calculator having +,-,*,/ operations, 
# using if else condition,
# taking the input from user for the operation to be performed

num1 = float(input("enter first number"))
num2 = float(input("enter second number"))

operation = input("enter operation +,-,* or / to be performed")

if operation == '+':
    result = num1 + num2
elif operation == '-':
    result = num1 - num2
elif operation == '*':
    result = num1 * num2
elif operation == '/':
    if num2!= 0:
        result = num1 / num2
    else:
        print("division by zero error")
else:
    print("invalid operation")

print(f"result is {result}")
```

```python
# determine ticket price based on age and is_student
age = int(input("enter age"))
is_student = input("Are you a student enter y/n")

if age <= 5:
    price = "free"
elif age <=12:
    price = "$10"
elif age <=17:
    if is_student == 'y':
        price = "$12"
    else:
        price = "$15"
elif age <= 64:
    if is_student == 'y':
        price = "$18"
    else:
        price = "$25"
else:
    price = "$20"

print("ticket price is: ",price)

```
