```python
# logical operators -> And , Not,or

# And -> when both conditions are true then only result is true
# or -> when atleast one condition is True, result is True
# not -> negates the initial value

x = True
y = True
print(x and y)

z = False
print(x and z)

print("OR condition")

w = False
print(x or y)
print(y or z)
print(z or w)
```
