```python
import os

# create new directory
new_directory = "package"

os.mkdir(new_directory)
print(f"directory {new_directory} created")
```

```python
# . indicates that we are presently looking at current working directory

items = os.listdir('.')
print(items)
```

```python
# joining paths
dir_name = "folder"
file_name = "file.txt"
full_path = os.path.join(dir_name,file_name)
print(full_path)
```

```python
full_path = os.path.join(os.getcwd(),dir_name,file_name)
print(full_path)
```

```python
# to check if the path exist
file = "dumm.txt"
if os.path.exists(file):
    print(f"file {file} exist")
else:
    print(f"file {file} does not exist")
```

```python
# checking if path is file or directory

path = "love1.txt"

if os.path.isfile(path):
    print(f"{path} is a file")
elif os.path.isdir(path):
    print(f"{path} is a directory")
else:
    print(f"{path} is neither a directory nor file")
```

```python
# getting absolute path

relative_path = "love.txt"
absolute_path = os.path.abspath(relative_path)
print(absolute_path)
```
