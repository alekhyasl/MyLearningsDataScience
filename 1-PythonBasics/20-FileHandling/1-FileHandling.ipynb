```python
f = open("funny.txt",'r',  encoding="utf-8") # opening the file in read mode
```

```python
for line in f:
    print(line)

f.close() # deallocates the resources
```

```python
f = open("funny.txt",'r',  encoding="utf-8") # opening the file in read mode

for line in f:
    print(line.strip()) # strip removes new line character

f.close()
```

```python
# here using with automatically closes the file as this instance
with open("funny.txt",'r', encoding="utf-8") as f:
    content = f.read()
    print(content)
  
```

```python
with open("funny.txt",'r',encoding="utf-8") as f:
    lines = f.readlines()
    for line in lines:
        print(line)
```

```python
with open("love.txt",'w') as f1: # creates the file and open in write mode
    f1.write("I love python\n")
    f1.write("I love coding\n")
```

```python
with open("love.txt",'w') as file: # here the content will be overridden
    file.write("Hello world!")
    file.write("this is new line")
```

```python
with open("love.txt",'a') as f2:  # opening in append mode will append text to file
    f2.write("I love golang\n")
    f2.write("I love ML\n")
```

```python
with open("love.txt",'a') as f3:
    f3.writelines("I love to learn \nI love to implement the things learnt\n")
```

```python
with open("love.txt",'w') as f4:
    f4.writelines([
        "I love coding\n",
        "In love learn to code\n",
        "I love to code inn different ways\n"
    ])
```

```python
# working with binary files
# binary files denoted with .bin
# mode -> wb -> write byte
# rb -> read byte

data = b'\x00\x01\x02\x03\x04' # hexadecimal data \x

with open("example.bin",'wb') as f:
    f.write(data)
```

```python
with open("example.bin",'rb') as f:
    content = f.read()
    print(content)
```

```python
# read the content from source file and write to destination file

with open("funny.txt",'r') as f:
    content = f.read()

with open("destination.txt",'w') as f:
    f.write(content)
```

```python
# count the number of lines , words  and characters 

with open("funny.txt",'r') as f:
    lines  = f.readlines()
    line_count = len(lines)
    word_count = sum(len(line.split()) for line in lines)
    char_count = sum(len(line) for line in lines)

print("number of lines: ",line_count)
print("number of words: ",word_count)
print("number of characters: ",char_count)
```

```python
# writing and then reading a file
# w+ mode is used for both reading and writing 
# if the file exists it overwrites else it creates one

f= open("love.txt",'+w')
f.write("hello from file handling\n")
f.write("this is new line\n")

# now when you try to perform read operation , the cursor is already at the end 
# so reading cannot be done till the cursor moves to the beginning

f.seek(0) # -> moving cursor to the beginning

# now reading the content

content = f.read()
print(content)
```
