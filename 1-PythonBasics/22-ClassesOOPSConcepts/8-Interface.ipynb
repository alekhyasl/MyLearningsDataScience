```python
# In python we say interface as Abstract base class
# Polymorphism with Abstract base class

# Abstract base classes(ABC) are used to define methods for group of related objects
# they enforce the derived classes to implement particular methods to promote
# consistency across different implementations
```

```python
from abc import ABC,abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass
```

```python
class Car(Vehicle):
    def start_engine(self):
        return "Car engine started"
```

```python
class MotorCycle(Vehicle):
    def start_engine(self):
        return "Motor Cycle engine started"
```

```python
car = Car()
motor = MotorCycle()
print(car.start_engine())
print(motor.start_engine())
```

```python
def start_engine(obj):
    print(obj.start_engine())
```

```python
start_engine(car)
start_engine(motor)
```
