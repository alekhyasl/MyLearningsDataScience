```python
virat = {
'fname' : "virat",
'lname' : "Kohli",
'scores' : [],
'birth_year' : 1988

}
```

```python
# appending scores of virat
virat['scores'].append(50)
virat['scores'].append(100)
virat['scores'].append(5)
```

```python

def get_avg_scores(player):
    return sum(player['scores'])/len(player['scores'])
print(get_avg_scores(virat))
```

```python
from datetime import date

def get_age(player):
    today = date.today()
    age = today.year - player['birth_year']
    return age
```

```python

print(get_age(virat))
```
