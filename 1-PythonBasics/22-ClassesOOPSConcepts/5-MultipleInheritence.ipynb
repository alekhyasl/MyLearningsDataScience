```python
# Multiple Inheritance -> when a class inherits from more than one base class
# this is multiple inheritance
```

```python
# Base class 1
class Animal:
    def __init__(self,name,food_type):
        self.name = name
        self.food_type = food_type

    def eats(self):
        print(f"Animal eats {self.food_type}")

    def speak(self):
        print("sub class must implement this method")
```

```python
class Pet:
    def __init__(self,owner):
        self.owner = owner
```

```python
class Dog(Animal,Pet):
    def __init__(self,name,food_type,owner):
        Animal.__init__(self,name,food_type)
        Pet.__init__(self,owner)

    def speak(self):
        return f"{self.name} says woof"
```

```python
dog = Dog("Buddy","Chicken","Alekhya",)
print(dog.speak()) # executes method from child class
print(f"owner : {dog.owner}")
dog.eats()
```
