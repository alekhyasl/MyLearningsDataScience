```python
# Inheritence allows class to inherit attributes and methods from another class
# parent class
 
class car:
    def __init__(self,windows,doors, enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype

    def drive(self):
        print(f"the person drives {self.enginetype} car ")
```

```python
car1 = car(4,4,"petrol")
car1.drive()
```

```python
class Tesla(car):
    def __init__(self, windows, doors, enginetype,is_self_driving):
        super().__init__(windows, doors, enginetype)
        self.is_self_driving = is_self_driving

    def self_driving(self):
        print(f"Tesla supports self driving {self.is_self_driving}")
    
```

```python
model3=Tesla(4,4,"electirc",True)
model3.self_driving()
model3.drive()
```

```python
from datetime import date
# parent class
class Player:
    def __init__(self, fname, lname, byear):
        self.fname = fname
        self.lname = lname
        self.birth_year = byear

    def get_age(self):
        today = date.today()
        age = today.year - self.birth_year
        return age
```

```python
class TennisPlayer(Player):
    def __init__(self, fname, lname, byear):
        Player.__init__(self,fname,lname,byear)
        self.aces = []

    def get_avg_aces(self):
        return sum(self.aces)/len(self.aces)
```

```python
saina = TennisPlayer("saina","agarwal",1994)
print(saina.fname)
print(saina.get_age())
```

```python
class CricketPlayer(Player):
    def __init__(self, fname, lname, byear,team):
        super().__init__(fname,lname,byear)
        self.team = team
        self.scores = []

    def get_avg_score(self):
        return sum(self.scores)/len(self.scores)

    def add_score(self,score):
        self.scores.append(score)
```

```python
rohit = CricketPlayer("rohit","sharma",1990,"India")
print(rohit.team)
print(rohit.lname)
print(rohit.get_age())
rohit.add_score(45)
```
