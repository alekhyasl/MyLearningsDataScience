```python
# Polymorphism - a way to perform a single action in different forms
# Polymorphism is achieved with help of method overriding and interfaces

# Method overriding allows the child class to implement the method that is
# already present in parent class
```

```python
# base class
class Animal:
    def speak(self):
        return "sound of the animal"
```

```python
# derived class
class Dog(Animal):

    def speak(self):
        return "woof woof!"
```

```python
class Cat(Animal):

    def speak(self):
        return "Meow!"
```

```python
dog = Dog()
print(dog.speak())
cat = Cat()
print(cat.speak())
```

```python
def animal_speaks(animal):
    print(animal.speak())

animal_speaks(dog)
```
