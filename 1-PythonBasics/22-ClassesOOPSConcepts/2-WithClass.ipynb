```python
class Dog:
    def __init__(self,name,age):
        self.name = name
        self.age = age

    def bark(self):
        print(f"{self.name} says woof")

dog1 = Dog("buddy",3)
print(dog1)
print(dog1.name)
print(dog1.age)
dog1.bark()
```

```python
from datetime import date
class CricketPlayer:
    # __init__ is the constructor where we can define attributes
    team_size = 11
    # constructor
    def __init__(self,fname,lname,byear,team):
        self.fname = fname
        self.lname = lname
        self.birth_year = byear
        self.scores =  []
        self.team = team
# defining instance methods
    def add_score(self,score):
        self.scores.append(score)

    def get_avg_score(self):
        return sum(self.scores)/len(self.scores)

    def get_age(self):
        today = date.today()
        age = today.year - self.birth_year
        return age
     # __str__, __lt__, __repr__,__eq__ ,__add__, __sub__ ,__mul__,
     #  __truediv__,__gt__
    # these are some of the magic methods or dunder methods which start with __
    # they are used to override the functionality of the object
    # so that you can define the custom functionality
    # for example when we do print(object), it prints the
    # memory location the object is pointing to
    # by using def __str__ function you can customise the function
    def __str__(self):
        return f"{self.fname} {self.lname} a cricket player from {self.team}"

    def __repr__(self):
        return f"Person(name={self.fname},age ={self.birth_year})"

    def __lt__(self, other): # this method is overloading the less than operator
        self_avg_scores = self.get_avg_score()
        other_avg_scores = other.get_avg_score()
        return self_avg_scores < other_avg_scores

    def __eq__(self, other): # this method is overloading equal to operator
        self_age = self.get_age()
        other_age = other.get_age()
        return self_age == other_age
```

```python
# when u create an object it calls the constructor first
virat = CricketPlayer("virat","kohli",1988,"India")
virat.add_score(67)
virat.add_score(89)
virat.add_score(231)
print(virat.fname)
print(virat.lname)
print(virat.birth_year)
print(virat.team)
print(virat.get_avg_score())
print(virat.get_age())
print(virat)
print(repr(virat))
```
