```python
# Encapsulation - is a concept of wrapping data(variables)  and methods(functions)
# together as a single unit. It restricts direct access to some of the object
# components which helps in preventing accidental interference and misuse of data

# Encapsulation with getter and setter methods
# public , protected and private access variables
```

```python
class Person:
    def __init__(self,name,age):
        self.name = name # public variables -> can be accessed outside class
        self.age = age # public variables
```

```python
def get_name(obj):
    return obj.name
```

```python
person = Person("Alekhya",31)
print(person.name)
print(get_name(person))
```

```python
print(dir(person)) 
# we can see age and name in the list as they are public variables
```

```python
# If we do not want to access/change variables outside the class , 
# we can do as below

class Student:
    def __init__(self,name,age,gender):
        self.__name = name  # private variables -> can't be accessed outside class
        self.__age = age    # private variables
        self.gender = gender
```

```python
student = Student("Alekhya",31,"Female")
print(dir(student)) # here the private variables are displayed as
# _Student__age , _Student__name

```

```python
print(student.__name) # gives error as it is accessible inside class only
```

```python
def getname(obj): # able to access name from outside class as it is public
    return obj.__name
```

```python
print(getname(student)) # gives an error saying __name attribute is not defined
# as __name is a private variable

# so private variable cannot be accessed outside the class, and cannot
# be accessed from the derived class
```

```python
# protected variable

class People:
    def __init__(self,name,age,gender):
        self._name = name  # protected variables -> can be accessed in
        # derived classes but not outside other classes
        self._age = age    # protected variables
        self.gender = gender
```

```python
class Employee(People):
    def __init__(self,name,age,gender):
        super().__init__(name,age,gender)
```

```python
emp = Employee("alekhya",32,"Female")
print(emp._age)
```

```python
# Encapsulation with getter and setter methods

class Members:
    def __init__(self,name,age,gender):
        self.__name = name
        self.__age = age
        self.gender = gender

    def get_name(self):
        return self.__name
    def get_age(self):
        return self.__age
    def set_name(self,name):
        self.__name = name
    def set_age(self,age):
        if age <0:
            print("age cannot be negative")
        else:
            self.__age = age
```

```python
mem = Members("alekhya",32,"F")
print(mem.get_age())
print(mem.get_name())
mem.set_name("chaitanya")
print(mem.get_name())
mem.set_age(34)
print(mem.get_age())
```
