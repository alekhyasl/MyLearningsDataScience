```python
class Shape:
    def area(self):
        return "The area of figure"
```

```python
class Rectangle(Shape):

    def __init__(self,length,width):
        self.length = length
        self.width = width

    def area(self):
        return f"the area is {self.length * self.width}"
```

```python
class Circle(Shape):
     def __init__(self,radius):
         self.radius = radius

     def area(self):
         return f"the area is {3.14 * self.radius * self.radius}"
```

```python
def area(obj):
    print(f"the area is {obj.area()}")
```

```python
rect = Rectangle(2,3)
cir = Circle(3)
area(rect)
area(cir)
```
