```python
# class variables -> also called static variables, shared across
# all instances of class

class Student:
    school_name = 'ABC School'

    def __init__(self,name,roll_no):
        self.name = name
        self.roll_no = roll_no
```

```python
s1 = Student("emma",2)
s2 = Student("hebba",3)
```

```python
print(s1.school_name)
print(Student.school_name)
```

```python
s1.school_name = "XYZ school"
```

```python
print(s1.school_name)
print(Student.school_name)
print(s2.school_name)
```

```python
# class variables in inheritance

class Course:
    course = "Python"

class Student(Course):

    def __init__(self,name):
        self.name = name

    def show_Student(self):
        print(f"{self.name} : {Student.course}")
```

```python
s1 = Student("Emma")
```

```python
s1.show_Student()
print(s1.course)
```

```python
Student.course = "AI"
print(s1.course)
print(Course.course)
```
