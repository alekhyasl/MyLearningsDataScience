```python
d1 = {'name' : 'alekhya','age' : 32,'gender' : "F"}
d2 = d1
print(d1)
print(d2)

```

```python
d1['name'] = 'alekhay2'
print(d1)
print(d2)
# here the change in value of d1 effects the value of d2 as d2 = d1
# references to d1 and does not create a new dictionary
# so the change in d1 effects d2 as well
```

```python
d2['name'] = 'alekhya3'
print(d1)
print(d2)
```

```python
# shallow copy
# You create a new dictionary object, but its values are not fully copied.
# If a value is mutable (like a list, another dict, set), 
# both dictionaries still share the same inner objects

d1 = {'name' : 'alekhya','age' : 32,'gender' : "F"}
d2 = d1.copy()

print(d1)
print(d2)


```

```python
d1['name'] = 'alekhay2'
print(d1)
print(d2)
```

```python
d2['name'] = 'alekhya3'
print(d1)
print(d2)
```

```python
# if we have mutable objects like list, set and dictionary in the dict
# both dictionaries still share the same inner objects.

d1 = {'a':1,'b':[1,2,3,4]}
d2 = d1.copy()
print(d1)
print(d2)
```

```python
d1['b'].append(5)
print(d1)
print(d2)

d1['a'] = 99
print(d1)
print(d2)
```

```python
# same shallow copy concept applies for list,set as well
v1 = ["sri",123,'alekhya']
v2 = v1.copy()
v1[1] = 786
print(v2)
print(v1)
```

```python
# accessing elements in nested dictionary
contact_dict = {
    'rachal' : {'phone' : 6787655434 , "address" : "Bangalore"},
    'swapna' : {'phone' : 6787655894 , "address" : "Mumbai"}
}

print(contact_dict['swapna'])
print(contact_dict['rachal']['address'])
```

```python

for contact in contact_dict:  # prints key values
    print(contact)
```

```python
for contact, value in contact_dict.items(): # prints both keys and values
    print(contact,value)
```

```python
print(contact_dict.keys())
print(contact_dict.values())
```

```python
# dictionary comprehension

squares = { i: i**2 for i in range(10)}
print(squares)
```

```python
# find squares of even numbers 
# condition dictionary comprehension

{i :i**2 for i in range(10) if i%2 == 0}
```

```python
# use dictionary to find frequency of elements in list
numbers = [1,2,2,3,3,3,4,4,4,4]
d1 = {}
for num in numbers:
    if num in d1.keys():
        d1[num] = d1[num] + 1
    else:
        d1[num] = 1

print(d1)
```
