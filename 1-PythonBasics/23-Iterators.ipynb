```python
# Iterators allow efficient looping and effective memory management
# they help in accessing the elements sequentially
my_list = [1,2,3,4,5,6,7]
for i in my_list:
    print(i)
```

```python
print(type(my_list))
```

```python
# creating an iterator using iter keyword

iterator = iter(my_list)

print(type(iterator)) # it is a list_iterator
```

```python
print(iterator) # uses lazy loading as it does not give the elements in 
# list iterator, it just gives the location of the elemnts
# when we iterate using next then the elements are loaded

```

```python
next(iterator)
```

```python
try:
    print(next(iterator))
except StopIteration:
    print("No elements in the iterator")
```

```python
str = "hello"
str_iterator = iter(str)
print(type(str_iterator))
for i in range(len(str)):
    print(next(str_iterator))
```

```python
prices = []

with open("pricing.txt",'r') as f:
    for line in f: # here all lines in file will not be loaded in memory at once, one line after other is loaded
        ticker,exchange,price,date=line.split('|')
        prices.append({'ticker':ticker,'exchange_listed':exchange,'price':float(price),'date':date})

print(prices[:3])
```

```python
# custom remote control

class RemoteControl:
    def __init__(self):
        
        self.channels = ['CNN',"NDTV",'BBC','HBO']
        self.index = -1
       
    def __next__(self):
        self.index += 1
        if self.index == len(self.channels):
            raise StopIteration
        return self.channels[self.index]
    def __iter__(self):
        return self
```

```python
rc = RemoteControl()
# here rc acts as an iterator as we used __next__ and __iter__ methods which means
# it is an iterator and loops through using next
# it prints channel names becoz th __next__ is returning channels
for i in rc:
    print(i)
```
