```python
def is_even_odd(number):
    """ This function finds if a number is even or odd"""
    if number%2 == 0:
        return "number is even"
    else:
        return " number is odd"


# calling function
    

print(is_even_odd(78))
print(is_even_odd(79))
```

```python
expense_sergey = [30,45,70,90]
expense_sundar = [40,23,10,85]
```

```python
def find_total(expenses):
    '''

    :param expenses: list containing numbers
    :return: total sum of all numbers in input list
    '''
    total = 0
    for expense in expenses:
        total += expense
    return  total
```

```python
sergey_total_expense = find_total(expense_sergey)
sundar_total_expense = find_total(expense_sundar)
```

```python
print(f"sergey total expense is {sergey_total_expense}")
print(f"sundar total expense is {sundar_total_expense}")

```

```python
# find cylinder volume

def find_cylinder_volume(radius,height):
    '''
    This function finds the volume of a cylinder given radius and height 
    '''
    volume = 3.14 * (radius ** 2) * height
    return volume

radius = 10
height = 7
v = find_cylinder_volume(radius,height)
print(f"cylinder volume is : {v}")
```

```python
# positional arguments -> using them we can change the order of arguments

v = find_cylinder_volume(height= height,radius=radius)
print(f"cylinder volume is : {v}")
```

```python
# assigning default value to parameters

def find_cuboid_volume(length,breadth,height= 3):
    '''

    :param length: length measurement
    :param breadth: breadth measurement
    :param height: height measurement
    :return:
    '''
    volume = length * breadth * height
    return  volume
volume = find_cuboid_volume(3,4)
print(f"volume of cuboid is : {volume}")

volume = find_cuboid_volume(3,4,4)
print(f"volume of cuboid is : {volume}")
```
