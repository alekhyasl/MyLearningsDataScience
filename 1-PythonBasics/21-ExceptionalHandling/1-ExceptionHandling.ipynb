```python
# Exception handling -> when any error occurs while program execution then
# program terminates at that point and throws the error
# to terminate execution in meaningful way we do exceptional handling using
# try except block

# in case the program executes without any error , except block is not executed
# in case program faces error it goes to except block executes the code in it
# and then continues the execution of rest of the code

# Exception occurs when an error is encountered during program exceution

# A variable exists outside the try block only if its assignment
# executed successfully.
```

```python
try:
    a=b
except:
    print("variable is not assigned")
```

```python
try:
    a=b
except NameError as ex:
    print(ex)
```

```python
try:
    #2/0
    1/2
    a=b
    
except ZeroDivisionError as ex:
    print(ex)
    print("denominator should be Non zero")
except Exception as ex: # parent class exception should be writeen at end
    print(ex)
```

```python
x = input("enter first number")
y = input("enter second number")
d= 0
try:

    d = int(x)/int(y)
    "buddy age is "+9

except ZeroDivisionError as ze:
    print("Exception is handled and is a zero division error :",ze)

except TypeError as te:
    print("exception occured ",te)
    

print(f"division of {x} and {y} is {d} ")
```

```python
# try except else block
# if none of exception gets raised else block gets executed

try:
    x = int(input("enter a number"))
    resuilt = 10/x
except ZeroDivisionError as ex:
    print(ex)
except ValueError as ex:
    print(ex)
except Exception as ex:
    print(ex)
else:
    print(f"the result is {resuilt}")
```

```python
# finally -> if exception is thrown or not finally always gets exceuted
file = None
try:
    file = open("text.txt", "r")
    content = file.read()
    print(content)
except FileNotFoundError as err:
    print("file not found: ",err)
else:
    print("file content read sucessfully")

finally:
    if file:
        file.close()
    else:
        print("file is already closed")
```
