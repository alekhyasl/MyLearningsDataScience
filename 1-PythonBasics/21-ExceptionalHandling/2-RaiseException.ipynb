```python
class InsuficientFunds(Exception):
    pass
```

```python
balance = 0
```

```python
def deposit(amount):
    global balance # global variable is available over entire py file
    if amount < 0:
        raise ValueError("Amount must be positive")
    else:
        balance += amount
        print(f"available balance is {balance}")
```

```python
# raising custom exception
def withdraw(amount):
    global balance
    if amount>balance:
        raise InsuficientFunds(f"No enough funds. current balance is {balance}")
    else:
        balance -= amount
        print(f"available balance is {balance}")
```

```python
deposit(10)
```

```python
deposit(-1)
```

```python
deposit(2)
```

```python
withdraw(8)
```

```python
withdraw(10)
```
