```python
# tuples are ordered collection of items which are immutable
# creating empty tuple

tup = tuple()
print(type(tup))
print(tup)

empty_tup = ()
print(empty_tup)
print(type(empty_tup))
```

```python
point = (6,7)
# accessing tuple elements
print(point[0])
print(point[1])
print(type(point))

```

```python
point[1] = 0   # tuples are immutable so cannot assign value
```

```python
point_3d = (5,6,7)

# converting list to tuple and vice versa

lst = [1,2,3,4,5]
print(tuple(lst))
tup = (7,8,9,10,11)
print(list(tup))
```

```python
# tuples can be heterogeneous

tup1 = (1,True,"Alekhya")
print(tup1)
```

```python
print("slicing for tuples")
numbers = (1,2,3,4,5,6)
print(numbers[2])
print(numbers[-1])
print(numbers[1:3])
```

```python
print("concatenation of tuples")
print(numbers+tup1)

```

```python

print(tup1 * 3) # elements repeats for 3 times
print(tup1)
```

```python
print(tup1.count(1)) # prints number of times 1 is present in tup1
# here it prints 2 as in tuple True is treated as 1
```

```python
print(tup1.index(True)) # returns the index where element is present
# here it prints 0, as True is 1 in python it returns 0 index
```

```python
# packing a tuple

packing_tuple = 1,True,"Alekhya","Kuchipudi"
print(packing_tuple)
type(packing_tuple)
```

```python
# unpacking a tuple
s_no,married,f_name,l_name=packing_tuple
print(s_no,f_name,l_name)
```

```python
# unpacking a tuple using *

num = (1,2,3,4,5,6,7,8)

print("unpacking with *")
first,*middle,last = num
print(first)
print(middle)
print(type(middle))
print(last)
```

```python
nested_tuple = ([1,2,3],(1,"Alekhya","kuchipudi",("Austin","Texas",78749)))
print(nested_tuple[1])
print(nested_tuple[1][3][1]) # prints Texas
```

```python
def find_pe_pb(price,eps,book_value):
    pe = price/eps
    pb = price/book_value
    return pe,pb

pe_ratio , pb_ratio = find_pe_pb(100,2,4)
print(pe_ratio)
print(pb_ratio)
```

```python
# storing names and phone numbers using list of tuples

contacts = [('rechal',4567876543),('monica',8987565432),('joey',9823111456)]
```

```python
# print monica phone number

for contact in contacts:
    if contact[0] == 'monica':
        print(f"monica phone number is {contact[1]}")
```
