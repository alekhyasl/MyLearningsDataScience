In python there are different data types like int , str, float, bool
we need not declare the data type to variables before assigning the values in python as we did in java and c++
python is dynamically typed language 
we can change the data type assigned to varible at any point of time

```python
can = 10
type(can)
```

```python
foo = "jalebi"
foo
```

```python
bar = foo
bar
```

```python
id(foo)
```

```python
id(bar) # here we can see bar is pointed to jalebi
```

```python
bar = "soda"
id(bar) # now id got changed as it is pointing to different value
```

```python
foo # still foo is pointing to jalebi
```

```python
id(foo)
```
