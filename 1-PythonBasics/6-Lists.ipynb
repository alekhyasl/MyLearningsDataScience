```python
# list comprehension
# syntax : [expression for item in iterable]
# with conditional logic : [expression for item in iterable if condition]
# nested list comprehension: 
# [expression for item1 in iterable for item2 in iterable ]

# list comprehension makes program faster and compact code
[x**2 for x in range(10)]
```

```python
# find cube of certain numbers using list comprehension

squares = [i**3 for i in range(1,15)]
print(squares)
```

```python
# find list of squares of even numbers

even_squares = [i**2 for i in range(1,15) if i % 2 == 0]
print(even_squares)
```

```python
# do the pairing for the items of lst1 and lst2

lst1 = [1,2,3,4]
lst2 = ['a','b','c','d']

[(i,j) for i in lst1 for j in lst2]
```

```python
# find length of words in list using list comprehension

words = ['hello','world','python','list','comprehension']

[len(word) for word in words]
```

```python
price = [100,120,80,50]
earnings_per_share = [9,15,40,2]
# cal p/e ratio using list comprehension

pe = [p/e for p,e in zip(price,earnings_per_share)]
print(pe)
```

```python
# if any of earnings per share is 0 , it has to be handled

[p/e if e!=0 else -1 for p,e in zip(price,earnings_per_share)]
```

```python
# apply 10 % discount if price > 100

prices = [120,99,150,80,200]

[price*0.9 if price> 100 else price for price in prices ]
```
