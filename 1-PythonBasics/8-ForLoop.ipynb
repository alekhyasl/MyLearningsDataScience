```python
# find total expenses
expenses = [100,240,340,320,400]
total = 0
for expense in expenses:
    total = expense + total

print(f"the total expense is {total}")
```

```python
total = 0
for i in range(len(expenses)):
    print(f"Month {i+1} : {expenses[i]}")
    total += expenses[i]

print(f"the total expense is {total}")
```

```python
# enumerate  loops over an iterable and 
# automatically get an index (counter) alongside each item
total = 0
for i , expense in enumerate(expenses):
    print(f"Month {i+1} : {expense}")
    total += expense

print(f"the total expense is {total}")
```
