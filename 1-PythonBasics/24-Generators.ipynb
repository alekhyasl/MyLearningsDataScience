```python
# Generators are sub class of iterators
# Generators are simple way to create iterators 
# uses yield keyword to produce series of values lazily
```

```python
def square(n):
    for i in range(n):
        return i ** 2

square(3)
```

```python
def square(n):
    for i in range(n):
        yield i **2

square(3)
```

```python
for i in square(3):
    print(i)
```

```python
'''
yield pauses the function, returns a value, and remembers where it left off
so it can continue later.
return ends the function; yield suspends it.

When Python hits a yield inside a function:

It produces the yielded value to the caller.

It freezes the function’s local variables and instruction pointer.

It pauses execution instead of exiting.

When the caller asks for the next value (next() or a for loop), 
the function resumes exactly after the last yield.
'''
```

```python
# Practical examples of using generators
# used for reading large files as they allow you to process one 
# line at a time without loading entire file to memory

def read_large_file(file_path):
    with open(file_path,'r') as f:
        for line in f:
            yield line

gen = read_large_file("large_file.txt")
for line in gen:
    print(line.strip())





```

```python
# fibnocci series with generator
def fibnocci():
    a,b=0,1
    while True:
        yield a
        a,b = b,a+b
      

f_gen = fibnocci()

    
# 0,1,1,2,3,5,8,13
```

```python

print(next(f_gen))
print(next(f_gen))
```
