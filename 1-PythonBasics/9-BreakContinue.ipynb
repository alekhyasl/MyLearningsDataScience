```python
monthly_sales = [37,45,36,33,89,10,12,45]

threshold = 35
```

```python
for sales_amount in monthly_sales:
    if sales_amount<threshold:
        print(f" monthly sales amount {sales_amount} is less than threshold")
        break
    else:
        print(f" monthly sales amount {sales_amount} is greater than threshold")
```

```python
monthly_sales = [37,45,36,33,89,10,12,45]
months = ["jan","feb","march","April","may","june","July","Aug"]
threshold = 35
```

```python
for sales_amount,month in zip(monthly_sales,months):
    if sales_amount<threshold:
        print(f" monthly sales amount {sales_amount} is less than threshold in {month}")
        break
    else:
        print(f" monthly sales amount {sales_amount} is greater than threshold in {month}")
```

```python
# Continue

# print odd numbers from 1 to 20
for i in range(1,20):
    if i % 2 == 0:
        continue
    print(i)
```

```python
# While loop
# print numbers from 1 to 10

n= 0
while n<=10:
    print(n)
    n += 1
```

```python
# pass statement
# we use pass when we difine the function and do nothing inside it

for i in range(5):
    if i == 4:
        pass
    print(i)
```

```python
# nested for loop

for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}")
```

```python
products = ["iPhone","iPad","Macbook"]
regions = ["USA","China","India"]
revenue = [32,30,24,28,22,19,16,10,3]
i = 0
for product in products:
    for region in regions:
        print(f"revenue of {product} in {region} is {revenue[i]}")
        i+=1
```

```python
# for else

for i in range(5):
    print(i)
else:
    print("for loop terminated")
```

```python
for num in range(2,101):
    for i in range(2,num):
        if num%i == 0:
            break
    else:
        print(num)
```
