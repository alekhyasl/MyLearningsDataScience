```python
# Sets -> unique collection of items that do not follow order
# help to eliminate duplicates elements,perform mathematical operation like
# union, intersection etc

my_set = {1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
empty_set = set()
print(empty_set)
print(type(empty_set))
```

```python
set1 = set([1,2,3,4,5,6])
print(set1)
print(type(set1))
```

```python
set2 = set([1,2,3,4,5,3,2,1,5,6,7])
print(set2)
```

```python
# basic set operations
# adding and removing elements

num_set = {1,2,3,4,5,6}
num_set.add(7)
print(num_set)
num_set.add(7)
print(num_set)
```

```python
num_set.remove(6)
print(num_set)
```

```python
num_set.remove(8) # gives error as element is not present
```

```python
num_set.discard(10) 
# discard does not give error even if the number is not present
num_set
```

```python
popped_ele = my_set.pop() # follows first in first out 
print(popped_ele)
print(my_set)
```

```python
# clear all elements in set
num_set.clear()
print(num_set)
```

```python
# set membership test -> checks weather the element is present in set or not
my_set = {1,2,3,4,5}
print(3 in my_set)
print(8 in my_set)
```

```python
# MAthematical operations

set1 = {1,2,3,4,5}
set2 = {4,5,6,7,8}

# union
union_set = set1.union(set2)
print(union_set)
```

```python
# intersection
intersection_set = set1.intersection(set2)
print(intersection_set)
```

```python
set1.intersection_update(set2) # updates set1
print(set1)
print(set2)
```

```python
set1 = {1,2,3,4,5}
set2 = {4,5,6,7,8}
difference_set = set1.difference(set2)
print(difference_set)
```

```python
set1.difference_update(set2)
print(set1)
```

```python
set1 = {1,2,3,4,5}
set2 = {4,5,6,7,8}
difference_set = set2.difference(set1)
print(difference_set)
```

```python
# symmetric difference -> unique elements from both sets are combined i.e 
# the common elements from both sets are removed

sym_diff_set = set1.symmetric_difference(set2)
print(sym_diff_set)
```

```python
# set methods

set1 = {1,2,3,4,5}
set2 = {4,5,6,7,8}

set1.issubset(set2)
```

```python
set1.issuperset(set2)
```

```python
set1 = {1,2,3,4}
set2 = {1,2}
set1.issuperset(set2)
```
