What is fact and dimension Table?

Suppose we have a sales data of the grocery store in which we have customer name, address , phone number, product, category, price
These details of customer name, address , phone number can be stored in a different table with a unique customer ID to eliminate redundant data , which is called Normalisation
Because of the Nornalisation redundant data can be eliminated, inconsistency of data can be eliminated
Here sales table is called fact table
customer table is called dimension table
similarly, the details of  product, category, price in sales table are stored in product table with productID column added
product table is called dimension table

What is star schema and snowflake schema?

In the above example of sales data.. sales table (fact table) is at center and customer,product table at sides which forms a star and is called star schema

suppose the dimension table points to other one or more dimension table it is called snowflake schema

we can have multiple fact tables
