```python
import pandas as pd
```

```python
indian_wether = pd.DataFrame({
    "city": ["mumbai","delhi","banglore"],
    "temperature": [32,45,30],
    "humidity": [80, 60, 78]
})
```

```python
print(indian_wether)
```

```python
usa_weather = pd.DataFrame({  "city": ["new york","chicago","orlando"],
    "temperature": [21,14,35],
    "humidity": [68, 65, 75]
})
```

```python
print(usa_weather)
```

```python
df = pd.concat([indian_wether,usa_weather])
print(df)
```

```python
df = pd.concat([indian_wether,usa_weather],ignore_index=True)
print(df)
```

```python
df = pd.concat([indian_wether,usa_weather],keys=['India','USA'])
print(df)
```

```python
df.loc['India']
```

```python
df.loc['USA']
```

```python
temperature_df = pd.DataFrame({'city': ["mumbai","delhi","banglore"],
                              'temperature' : [32,45,30]},
                              index=[0,1,2])
print(temperature_df)
```

```python
windspeed_df = pd.DataFrame({'city' : ["delhi","mumbai"],
                             'windspeed' : [7,12]},
                             index=[1,0])
print(windspeed_df)
```

```python
df = pd.concat([temperature_df,windspeed_df],axis=1)
print(df)
```

```python
pd.concat([temperature_df,windspeed_df])
```

```python
temp_df = pd.DataFrame({
    "city": ["new york","chicago","orlando"],
    "temperature": [21,14,35],
})
print(temp_df)
```

```python
humidity_df = pd.DataFrame({
    "city": ["chicago","new york","orlando"],
    "humidity": [65,68,75],
})
print(humidity_df)
```

```python
pd.merge(temp_df,humidity_df,on='city')
```

```python
df1 = pd.DataFrame({
    "city": ["new york","chicago","orlando", "baltimore"],
    "temperature": [21,14,35, 38],
})
df1
```

```python
df2 = pd.DataFrame({
    "city": ["chicago","new york","san diego"],
    "humidity": [65,68,71],
})
df2
```

```python
pd.merge(df1,df2,on='city',how='inner')
```

```python
pd.merge(df1,df2,on='city',how='left')
```

```python
pd.merge(df1,df2,on='city',how='right')
```

```python
pd.merge(df1,df2,on='city',how='outer')
```

```python
pd.merge(df1,df2,how='cross')
```
