```python
import pandas as pd

df_sales = pd.read_excel("linechart1.xlsx")

print(df_sales)
```

```python
# plotting the line chart

from matplotlib import pyplot as plt
# to fix the chart size
plt.figure(figsize=(12,4))
plt.plot(df_sales['Quarter'],df_sales['Fridge'],color = 'blue',label = 'Fridge')
plt.plot(df_sales['Quarter'],df_sales['Dishwasher'],color = 'green',label = "Dishwasher")
plt.plot(df_sales['Quarter'],df_sales['Washing Machine'],color = 'red',label = 'Washing Machine')
plt.title("Product Sales")
plt.ylabel("Revenue ($ mln)")
plt.xlabel("Financial Quarter")
plt.legend()
plt.show()
```

```python
# to plot the total revenue share of each poduct we need to get
# sum of sales  of each product
total_sales = df_sales[['Fridge','Dishwasher','Washing Machine']].sum()
print(total_sales)
print(total_sales.index)
# autopct controls how the percentage values are displayed on each slice.
# %1.1f%% -> displays one decimal value
# %1.2f%% -> displays two decimal values
plt.pie(total_sales,labels=total_sales.index,autopct='%1.2f%%',explode=(0.1,0,0),
        shadow=True, startangle=140)
plt.show()
```

```python
df_sales.plot(kind = 'bar',x='Quarter') 
# here x = 'Quarter indicates that Quarter has to be plotted on x axis
plt.xticks(rotation=45)
plt.ylabel('revenue in mln($)')
plt.title('product sales')
plt.show()
```

```python
df_score = pd.read_excel("Histograms.xlsx")
print(df_score)
```

```python
plt.hist(df_score['Blood_Sugar'])
plt.show()
```

```python
# histograms look bettter with seaborn than matplotlib
from matplotlib import pyplot as plt
import seaborn as sns
sns.histplot(df_score['Blood_Sugar'],kde=True)
plt.xlabel('Blood Sugar')
plt.ylabel('Count')
plt.title('Blood Sugar Range')
```

```python
# another way of plotting a histogram in seaborn 
sns.histplot(data=df_score,x = 'Blood_Sugar',kde=True)
plt.title("Blood Sugar Levels")
plt.show()
```
