```python
import numpy as np
# install numpy using "!pip install numpy" in git bash
# numpy is basically a numerical python
# benefits of numpy array over list
# 1. array occupies less memory than list
# because of following reasons
# list does not store the integers directly.
# Each element is a pointer to a separate int object in memory.
# So memory includes:
# - The list container
# - A pointer for each element
# - A full Python int object for each number

# A NumPy array stores values in contiguous memory,
# using a fixed data type like int32 or float64.
# No pointers.
# No Python objects. Just raw bytes.
```

```python
import time
ls1 = list(range(1000000))
ls2 = list(range(1000000))
start_time = time.time()
ls3=[x+y for x, y in zip(ls1,ls2)]
end_time = time.time()
print("time taken for list: ",end_time-start_time)
```

```python
n1 = np.arange(1000000)
n2 = np.arange(1000000)
start_time = time.time()
n3 = n1 + n2
end_time = time.time()
print("time taken for array: ",end_time-start_time)
```
