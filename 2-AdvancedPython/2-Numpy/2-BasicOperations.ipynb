```python
import numpy as np
```

```python
# attributes of numpy array
arr = np.array([[1,2,3],[4,5,6]])
print(arr)
print("shape of array ",arr.shape)
print("dimensions of array",arr.ndim)
print("size (number of elements)",arr.size)
print("data type ",arr.dtype)
print("items size(in bytes) ",arr.itemsize) # ecah element size in array
print("bytes for entire array ",arr.nbytes)

'''
Memory usage  of an array
Total memory = itemsize * number_of_elements
'''
```

```python
rev_q1 = np.array([10,12,9]) # revenue in Q1
```

```python
print(rev_q1)
print(rev_q1.shape) # prints shape
print(rev_q1.ndim) # prints dimension
print(rev_q1.nbytes) # number of bytes occupied by array
```

```python
rev = np.array([[10,12,9],[23,15,19]]) # revneue in Q1 and Q2
print(rev)
print(rev.ndim)
print(rev[1,0]) # access first month revenue in Q2
rev[1,0] = 18 # manipulation of an element in array
print(rev)
```

```python
print(rev.dtype) # prints data type of array
```

```python
# we can assign a data type to array
rev = np.array([[12,14,15],[18,19,23]],dtype= np.float64)
print(rev.dtype) # prints data type of array
```

```python
print(rev.nbytes) # prints the number of bytes occupied by array
print(rev.itemsize) # size of each item in array
print(rev.size) # prints size of array
```

```python
print(rev.shape)
print(rev.reshape(3,2))
print(rev.shape)
```

```python
rev1 = np.array([[6,2,3],[1,5,2]])
print(np.sort(rev1)) # here axis is not specified so by default it takes axis = -1
# so it sorts row wise

print(np.sort(rev1, axis=1)) # sorts column wise
```

```python
print(np.sort(rev,axis=None)) # array gets flattened if axis is None
```

```python
# np.ones and np.zero is used in Deep learning for initializing weights
print(np.zeros((2,3))) # create 2*3 array with zeros
print(np.ones((2,4)))  # create 2*3 array with ones
print(np.eye(3)) # the diagonal of matrix is 1 and rest are 0 -> identity matrix
```

```python
print(np.arange(20,30,2))# here 2 is step size
```

```python
#Returns evenly spaced samples, calculated over the interval
print(np.linspace(1,2)) # by default  50 samples are printed

print(np.linspace(2,3,10))
```

```python
# to make an array flat
arr1 = np.array([[1,2,3],[4,5,6]])
print(arr1)
print(np.ravel(arr1)) # creates view does not modify original array
print(arr1.ravel())
print(arr1)
```

```python
print(arr1.flatten())
print(arr1)
```

```python
# difference between ravel and flatten
# Use ravel() when you want speed and don’t need a guaranteed copy.
# Use flatten() when you want to ensure the result is independent of the
# original array
```

```python
a = np.arange(6).reshape(2, 3)
b = a.ravel()
b[0] = 99 # a gets modified
```

```python
print(a)
```

```python
a = np.arange(6).reshape(2, 3)
b = a.flatten()
b[0] = 99 # a is not modified
```

```python
print(a)
```

```python
a2 = np.arange(12)
a2 = a2.reshape(3,4)
print(a2)
```

```python
print(a2.min())
print(a2.max())
print(a2.sum(axis = 1)) # cal sum of each row
print(a2.sum(axis = 0)) # cal sum of each column
print(a2.std())
print(a2.mean())
```

```python
# numpy vectorised operation

arr1 = np.array([1,2,3,4,5])
arr2 = np.array([10,20,30,40,50])

# element wise addition
print("addition : ",arr1+arr2)

# element wise subtraction
print("subtraction: ",arr1-arr2)

# element with multiplication
print("multiplication: ",arr1 * arr2)

# element wise division

print("division: ",arr1/arr2)
```

```python
# universal functions -> functions that apply to the entire array

arr = np.array([2,4,6,8,9])
# square root
np.sqrt(arr)
```

```python
# exponential
np.exp(arr)
```

```python
np.sin(arr)
```

```python
# natual log
np.log(arr)
```
